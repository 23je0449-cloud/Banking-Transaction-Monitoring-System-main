# Banking Transaction Monitoring

Explore synthetic banking transactions, monitor key metrics, and review rule-flagged records. This is a portfolio learning project, not a production fraud detection system.

```
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
# If needed, change ROOT to the project folder
customers = pd.read_csv(ROOT / 'data/raw/customers.csv')
transactions = pd.read_csv(ROOT / 'data/raw/transactions.csv', parse_dates=['transaction_date'])
customers.head()
```

```
transactions['status'].value_counts()
```

```
successful = transactions[transactions['status'].eq('Success')]
print('Total transactions:', len(transactions))
print('Successful value:', round(successful['amount'].sum(), 2))
print('Average successful transaction:', round(successful['amount'].mean(), 2))
```

```
transactions['month'] = transactions['transaction_date'].dt.to_period('M').astype(str)
monthly = successful.assign(month=successful['transaction_date'].dt.to_period('M').astype(str)).groupby('month', as_index=False).agg(transaction_count=('transaction_id','count'), transaction_value=('amount','sum'))
monthly.head()
```
